```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('student_habits_performance.csv')
df.head()
```

```python
df.isnull().sum()
```

```python
# drop student_id,parent_education column
df.drop(['student_id','parental_education_level'],axis=1,inplace=True)
```

```python
df['social_media_hours'] = df['social_media_hours']+df['netflix_hours']
df.drop(['netflix_hours'],axis=1,inplace=True)
```

```python
df.info()
```

```python
df['gender'].value_counts()
```

```python
df['gender']=df['gender'].map({'Male':1,'Female':0,'Other':2})
df.info()
```

```python
df['part_time_job']=df['part_time_job'].map({'No':0,'Yes':1})
```

```python
df.info()
```

```python
df['extracurricular_participation']=df['extracurricular_participation'].map({'No':0,'Yes':1})
```

```python
df['diet_quality']=df['diet_quality'].map({'Poor':0,'Fair':1,'Good':2})
```

```python
df.isnull().sum()
```

```python
df['internet_quality']=df['internet_quality'].map({'Poor':0,'Average':1,'Good':2})
```

```python
df.isnull().sum()
```

```python
df.info()
```

```python
# check outliers
plt.figure(figsize=(12,8))
sns.boxplot(data=df)
plt.xticks(rotation=45)
plt.show()
```

```python
# store outliers columer in l list
outliers_columer=[]
for col in df.columns:
    Q1=df[col].quantile(0.25)
    Q3=df[col].quantile(0.75)
    IQR=Q3-Q1
    lower_bound=Q1-1.5*IQR
    upper_bound=Q3+1.5*IQR
    outliers=df[(df[col]<lower_bound) | (df[col]>upper_bound)]
    if len(outliers)>0:
        outliers_columer.append(col)
print(outliers_columer)
```

```python
for i in outliers_columer[:-1]:
    Q1=df[i].quantile(0.25)
    Q3=df[i].quantile(0.75)
    IQR=Q3-Q1
    lower_bound=Q1-1.5*IQR
    upper_bound=Q3+1.5*IQR
    df=df[(df[i]>=lower_bound) & (df[i]<=upper_bound)]
```

```python
for i in df.columns:
    sns.boxplot(x=df[i])
    plt.show()
```

```python
# remove oultiers in social_media_hours column
Q1=df['social_media_hours'].quantile(0.25)
Q3=df['social_media_hours'].quantile(0.75)
IQR=Q3-Q1
lower_bound=Q1-1.5*IQR
upper_bound=Q3+1.5*IQR
df=df[(df['social_media_hours']>=lower_bound) & (df['social_media_hours']<=upper_bound)]
```

```python
sns.boxplot(x=df['social_media_hours'])
plt.show()
```

```python
# k-means clustering
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
scaled_data=scaler.fit_transform(df)
```

```python
kmeans=KMeans(n_clusters=4,random_state=42)
kmeans.fit(scaled_data)
```
